# Music Session Splitting

### What does this notebook do?

1. Creating 10 30min batches each with the around 60 songs per batch
2. Balance out those session to have an equal distribution of genre

## 1. Create list of all song names in the dataset

```python
def create_genre_song_list(genre: str):
  acc = []
  for i in range (100):
    acc.append(f"{genre}." + "{:05d}".format(i) + ".wav")
  return acc
```

```python
# create list of genres

genres = ["blues", "classical", "country", "disco", "hiphop", "jazz", "metal", "pop", "reggae", "rock"]
songs = []
for genre in genres:
  songs.append(create_genre_song_list(genre=genre))
map_of_songs = dict(zip(genres, songs))
```
