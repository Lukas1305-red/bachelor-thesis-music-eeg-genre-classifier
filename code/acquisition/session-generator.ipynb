# Music Session Splitting

### What does this notebook do?

1. Creating 10 30min batches each with the around 60 songs per batch
2. Balance out those session to have an equal distribution of genre

## 1. Create list of all song names in the dataset

```python
def create_genre_song_list(genre: str):
  acc = []
  for i in range (100):
    acc.append(f"{genre}." + "{:05d}".format(i) + ".wav")
  return acc
```

```python
# create list of genres

genres = ["blues", "classical", "country", "disco", "hiphop", "jazz", "metal", "pop", "reggae", "rock"]
songs = []
for genre in genres:
  songs.append(create_genre_song_list(genre=genre))
map_of_songs = {}
for i in range (len(genres)):
  map_of_songs[genres[i]] = songs[i]
```

```python
map_of_songs
```

## 2. Create Sessions with Music Songs

```python
SESSION_COUNT = 20
SONGS_PER_SESSION = 50
SONGS_PER_SESSION_PER_GENRE = 5
```

```python
pointer = 0
session_list = []
for current_session in range (SESSION_COUNT):
  session_act = []
  for genre in genres:
    for song_in_genre in range (SONGS_PER_SESSION_PER_GENRE):
      session_act.append(map_of_songs[genre][pointer + song_in_genre])
  pointer += 5
  session_list.append(session_act)
```

```python
# Shuffle songs per session for mixed genres

import random
for session in session_list:
  random.shuffle(session)
```
