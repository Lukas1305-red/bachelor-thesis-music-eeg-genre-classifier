```python
import numpy as np

def filter_inf_nan_columns(df):
    """
    Filter DataFrame to show only columns that contain inf or NaN values
    
    Args:
        df: pandas DataFrame
    
    Returns:
        pandas DataFrame with only columns containing inf or NaN values
    """
    
    # Find columns with NaN values
    nan_columns = df.columns[df.isnull().any()].tolist()
    
    # Find columns with inf values
    inf_columns = df.columns[df.isin([np.inf, -np.inf]).any()].tolist()
    
    # Combine both lists and remove duplicates
    problematic_columns = list(set(nan_columns + inf_columns))
    
    # Return filtered DataFrame
    return df[problematic_columns]

def get_inf_nan_summary(df):
    """
    Get a summary of which columns have inf/NaN values and their counts
    
    Args:
        df: pandas DataFrame
    
    Returns:
        pandas DataFrame with summary statistics
    """
    
    summary = []
    
    for col in df.columns:
        nan_count = df[col].isnull().sum()
        inf_count = df[col].isin([np.inf, -np.inf]).sum()
        
        if nan_count > 0 or inf_count > 0:
            summary.append({
                'Column': col,
                'NaN_Count': nan_count,
                'Inf_Count': inf_count,
                'Total_Problematic': nan_count + inf_count,
                'Percentage': ((nan_count + inf_count) / len(df)) * 100
            })
    
    return pd.DataFrame(summary)

def get_problematic_column_names(df):
    """
    Get just the column names that have inf or NaN values
    
    Args:
        df: pandas DataFrame
    
    Returns:
        list of column names with inf/NaN values
    """
    
    nan_columns = df.columns[df.isnull().any()].tolist()
    inf_columns = df.columns[df.isin([np.inf, -np.inf]).any()].tolist()
    
    return list(set(nan_columns + inf_columns))

# Example usage:
"""
# Create sample DataFrame with some inf/NaN values
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [1, np.inf, 3, 4],
    'C': [1, 2, 3, 4],  # No problematic values
    'D': [np.nan, 2, -np.inf, 4]
})

# Filter to show only problematic columns
problematic_df = filter_inf_nan_columns(df)
print("Columns with inf/NaN values:")
print(problematic_df)

# Get summary
summary = get_inf_nan_summary(df)
print("\nSummary:")
print(summary)

# Get just column names
problem_cols = get_problematic_column_names(df)
print("\nProblematic column names:")
print(problem_cols)
"""
```

```python
def find_list_columns(df):
    """
    Find columns that contain lists
    
    Args:
        df: pandas DataFrame
    
    Returns:
        list of column names that contain lists
    """
    list_columns = []
    
    for col in df.columns:
        # Check if any value in the column is a list
        if df[col].apply(lambda x: isinstance(x, list)).any():
            list_columns.append(col)
    
    return list_columns
```
